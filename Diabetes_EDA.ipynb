```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('diabetes_risk_dataset.csv')
df.head()
```

```python
sns.countplot(x='diabetes_risk_category',data=df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='bmi',
    y='fasting_glucose_level',
    hue='diabetes_risk_category',
    palette='Set1'
)

plt.title("BMI vs Fasting Glucose Level by Diabetes Risk Category")
plt.xlabel("BMI")
plt.ylabel("Fasting Glucose Level")
plt.savefig('ScatterPlot.png')
plt.show()
```

```python
df=df.drop('Patient_ID',axis=1)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))

sns.countplot(
    x='diabetes_risk_category',
    data=df,
    palette='Set2'
)

plt.title("Distribution of Diabetes Risk Categories")
plt.xlabel("Diabetes Risk Category")
plt.ylabel("Count")
plt.savefig('CountPlot.png')
plt.show()
```

```python
plt.figure(figsize=(12,8))

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt=".2f"
)

plt.title("Correlation Heatmap of Health Features")
plt.savefig('HeatMap.png',bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.boxplot(
    x='diabetes_risk_category',
    y='bmi',
    data=df,
    palette='Set2'
)

plt.title("BMI Distribution Across Diabetes Risk Categories")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='bmi',
    bins=30,
    kde=True,
    color='skyblue'
)

plt.title("Distribution of BMI Values")
plt.xlabel("BMI")
plt.ylabel("Frequency")

plt.show()
```
