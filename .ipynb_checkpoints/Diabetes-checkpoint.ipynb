### Encoding

```python
x=df.select_dtypes(include=['object','category'])
print(x.columns)
```

```python
from sklearn.preprocessing import LabelEncoder

legen=LabelEncoder()
lephy=LabelEncoder()
lefam=LabelEncoder()
letarg=LabelEncoder()
```

```python
df['gender']=legen.fit_transform(df['gender'])
df['physical_activity_level']=lephy.fit_transform(df['physical_activity_level'])
df['family_history_diabetes']=lefam.fit_transform(df['family_history_diabetes'])
df['diabetes_risk_category']=letarg.fit_transform(df['diabetes_risk_category'])

df.head()
```

#### Train Test Split

```python
X=df.drop('diabetes_risk_category',axis=1)
y=df['diabetes_risk_category']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

#### Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

### Model

### 1.LogisticRegression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
model_log=LogisticRegression()
model_log.fit(X_train,y_train)
```

```python
y_pred1=model_log.predict(X_test)
y_pred1
```

```python
accuracy_score(y_test,y_pred1)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred1))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred1))
print("\nClassification Report:\n", classification_report(y_test, y_pred1))
```

### 2.KNN

```python
from sklearn.neighbors import KNeighborsClassifier
knn_model=KNeighborsClassifier(n_neighbors=3)
knn_model.fit(X_train,y_train)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred2))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred2))
print("\nClassification Report:\n", classification_report(y_test, y_pred2))
```

### 3.Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
dt_model=DecisionTreeClassifier()
dt_model.fit(X_train,y_train)
```

```python
y_pred3=dt_model.predict(X_test)
y_pred3
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred3))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred3))
print("\nClassification Report:\n", classification_report(y_test, y_pred3))
```

```python
train_acc = dt_model.score(X_train, y_train)
test_acc = dt_model.score(X_test, y_test)

print("Train Accuracy:", train_acc)
print("Test Accuracy:", test_acc)
```

#### HYPER PARAMETER TUNING

#### Logistic Regression

```python
from sklearn.model_selection import GridSearchCV
```

```python
# LOGISTIC REGRESSION
#--------------------------
lr_params={'C':[0.1,1,10]}
lr_grid=GridSearchCV(LogisticRegression(),lr_params,cv=5)
lr_grid.fit(X_train,y_train)
```

#### KNN

```python
# KNN

knn_param={
    'n_neighbors': range(1,31),  # Number of neighbors
    'weights':['uniform','distance'],  # Voting strategy
    'metric':['euclidean','manhattan','minkowski']    # Distance metric
}

knn_grid=GridSearchCV(KNeighborsClassifier(),knn_param,cv=5,scoring='accuracy',n_jobs=-1)
knn_grid.fit(X_train,y_train)
```

#### Decision Tree

```python
# DECISION TREE
#----------------------
dt_params={'max_depth':[2,4,6,None],'criterion':['gini','entropy']}
dt_grid=GridSearchCV(DecisionTreeClassifier(random_state=42),dt_params,cv=5)
dt_grid.fit(X_train,y_train)
```

#### Random Forest

```python
# RANDOM FOREST
#-----------------------

rf_params={'n_estimators':[10,50,100],'max_depth':[2,4,6,None]}
rf_grid=GridSearchCV(RandomForestClassifier(random_state=42),rf_params,cv=5)
rf_grid.fit(X_train,y_train)
```

#### Guassian

```python
gb_param = {'var_smoothing': np.logspace(0, -9, num=100)}

# 3. Run Grid Search
gb_grid = GridSearchCV(GaussianNB(), gb_param, cv=5,scoring='accuracy',verbose=1)
gb_grid.fit(X_train,y_train)
```

#### XGBoost

```python
# xgb
xgb_param={
    'n_estimators': [100,200,300],
    'learning_rate':[0.01,0.1,0.2],
    'max_depth':[3,5,7],
    'min_samples_split':[2,5,10]
}

xgb_grid=GridSearchCV(XGBClassifier(),xgb_param,cv=5,scoring='accuracy',n_jobs=-1)
xgb_grid.fit(X_train,y_train)
```

#### ADABoost

```python
ada_param = {
    'n_estimators': [50, 100, 200],
    'estimator':[None],
    'learning_rate': [0.01, 0.1, 1.0]
    
}
ada_grid = GridSearchCV(AdaBoostClassifier(), ada_param, cv=5, scoring='accuracy', n_jobs=-1)
ada_grid.fit(X_train, y_train)
```

#### GradientBoost

```python
gbc_param = {
    'n_estimators': [50, 100, 200],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [3, 4, 5, 6]
}

gbc_grid = GridSearchCV(GradientBoostingClassifier(),gbc_param, cv=5, scoring='accuracy', n_jobs=-1)
gbc_grid.fit(X_train, y_train)
```

#### Result

```python
models={'Logistic Regression': lr_grid,'KNN': knn_grid,'Decision Tree': dt_grid,'Random Forest': rf_grid,'GaussianNB':gb_grid,'XGBoost':xgb_grid,'ADABoost':ada_grid,'GradientBoost':gbc_grid}
for name, grid in models.items():
    best_model = grid.best_estimator_
    y_pred = best_model.predict(X_test)
    acc = accuracy_score(y_test,y_pred)
    print(f"{name}:")
    print(f' Best Params: {grid.best_params_}')
    print(f" Accuracy: {acc:.4f}\n")
```

#### From this it is clear that Logistic Regression Model is the best performing model
